```python
import matplotlib.pyplot as plt
import numpy as np

# Enable 3D plotting
from mpl_toolkits.mplot3d import Axes3D

# Generate 10 random integer values for A, B, C (e.g., between 1 and 10)
A = np.random.randint(1, 11, 1)
B = np.random.randint(1, 11, 1)
C = np.random.randint(1, 11, 1)

print(f"Coefficients: A={A[0]}, B={B[0]}, C={C[0]}")

# Generate 10 random points for x, y, z (e.g., between -10 and 10)
x = np.random.uniform(-10, 10, 10)
y = np.random.uniform(-10, 10, 10)
z = np.random.uniform(-10, 10, 10)

# Calculate the scalar value for each point: Ax + By - Cz
scalar_values = A[0] * x + B[0] * y - C[0] * z

# Create a 3D scatter plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot with colors based on scalar_values
scatter = ax.scatter(x, y, z, c=scalar_values, cmap='viridis', s=100, alpha=0.7)

# Add labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title(f'3D Scatter Plot of {A[0]}x + {B[0]}y - {C[0]}z')

# Add a color bar to show the mapping of scalar values to colors
cbar = fig.colorbar(scatter, ax=ax, pad=0.1)
cbar.set_label('Scalar Value (Ax + By - Cz)')

plt.show()
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('stopwords')
```

Stop words are common words in a language (like "the", "a", "is", "and") that are often removed from text before processing, as they usually don't carry significant meaning for text analysis tasks. NLTK (Natural Language Toolkit) provides a list of these common words for various languages.

```python
#printing the stopwords
print(stopwords.words('english'))
```

```python
#data preprosesing
news_data.isnull().sum()
```

```python
#replacing the missing value with null string
news_data=news_data.fillna('')
news_data['content']=news_data['author']+' '+news_data['title']
news_data.head()
```

```python
# separating feature and target her , Target->lavel and feature->content
X=news_data.drop(columns='label',axis=1)
Y=news_data['label']
```

```python
#steming is the process of reducing a word to its keyword
port_stem=PorterStemmer()

def stemming(content):
   stemmed_content=re.sub('[^a-zA-Z]',' ',content)
   stemmed_content=stemmed_content.lower()
   stemmed_content=stemmed_content.split()
   stemmed_content=[port_stem.stem(word) for word in stemmed_content if not word in stopwords.words('english')]
   stemmed_content=' '.join(stemmed_content)
   return stemmed_content
```

```python

   news_data['content']=news_data['content'].apply(stemming)
```

```python
X=news_data['content'].values
Y=news_data['label'].values
print(X)
```

```python

vectorizer=TfidfVectorizer()
vectorizer.fit(X)
X=vectorizer.transform(X)
print(X)
```
